```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from fontTools.diff import color

img1 = np.zeros((100, 100, 3), dtype="uint8")
img1[50][50][0] = 255
plt.axis('off')
plt.imshow(img1)
plt.show()
```

```python
#diagonal
height = 100
width = 100
img1 = np.zeros((height, width, 3), dtype="uint8")
for i in range(height):
    for j in range(width):
        if i == j:
            img1[i][j][0] = 106  # R - red
            img1[i][j][1] = 31  # G - green
            img1[i][j][2] = 191  # B - blue

plt.axis('off')
plt.imshow(img1)
plt.show()
```

```python
#diagonal
height = 100
width = 100
img1 = np.zeros((height, width, 3), dtype="uint8")
cv2.line(img1, (0, 0), (width, height), color=(106, 31, 191))
plt.axis('off')
plt.imshow(img1)
plt.show()
```

```python
#diagonal
height = 100
width = 100
img1 = np.zeros((height, width, 3), dtype="uint8")
cv2.line(img1, (0, height), (width, 0), color=(106, 31, 191))
plt.axis('off')
plt.imshow(img1)
plt.show()
```

```python
cv2.line(img1, (0, 0), (width, height), color=(106, 31, 191))
cv2.line(img1, (0, height), (width, 0), color=(106, 31, 191))
cv2.line(img1, (0, height // 2), (width, height // 2), color=(106, 31, 191))
cv2.line(img1, (width // 2, 0), (width // 2, height), color=(106, 31, 191))
# plt.axis('off')
plt.imshow(img1)
plt.show()
```

```python
cv2.circle(img1, (width // 2, height // 2), width // 3, color=(0, 255, 0), thickness=1)
plt.axis()
plt.imshow(img1)
plt.show()
```
