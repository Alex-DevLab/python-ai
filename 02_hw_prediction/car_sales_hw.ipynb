```python
import pandas as pd

df = pd.read_csv('../data/car_sales_cleaned1.csv')
df = df.drop_duplicates()
df = df.dropna()
n_unique = df['Model'].nunique()
v_counts = df['Model'].value_counts(normalize=True).head()
n_unique
v_counts
```

```python
import converter

mapper = converter.columns_mapper(['Gender', 'Color'], df)
df_converted = converter.convert_x(df, mapper)
X = df_converted[['Gender', 'Color']]
y = df['Model']
df_converted
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
prediction = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

score = accuracy_score(y_test,prediction)
score
```

```python
import joblib

joblib.dump(model,'../models/gender_color__model.joblib')
joblib.dump(mapper, '../mappers/gender_color__mapper.joblib')
```
