```python
import numpy as np

data_set = np.array([
    [0.3, -1.2, 0.5, 0.9],
    [0.7, -2.7, 0.6, 0.1],
    [0.3, 0.01, 0.47, 2],
    [0.5, 2.2, 0.11, 3.0],
    [0.12, -0.7, 0.15, 0.25]
])
vec = np.array([0.6, -1.5, 0.3, 0.2])
vec_len = np.linalg.norm(vec)
vec_len
vec_norm = vec / vec_len
vec_norm
```

```python
vec_norm_len = np.linalg.norm(vec_norm)
vec_norm_len
```

```python
# data_set_len = np.linalg.norm(data_set)
# data_set_len
data_set_lengths = np.linalg.norm(data_set, axis=1, keepdims=True)
data_set_lengths
```

```python
data_set_norm = data_set / data_set_lengths
data_set_norm
```

```python
data_set_norm_len = np.linalg.norm(data_set_norm, axis=1, keepdims=True)
data_set_norm_len
```

```python
similarities = data_set_norm @ vec_norm
similarities
```

```python
best_index = np.argmax(similarities)
best_index
```
